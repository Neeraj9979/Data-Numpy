```python
import numpy as np
```

```python
arr1=np.array([1,2,3,4,5])
print(arr1,arr1.dtype)
arr2=np.array([1.2,2.3,3.4,4.5,5.5])
print(arr2,arr2.dtype)
arr3=np.array(["Hi","Hello","How","are","you"])
print(arr3,arr3.dtype)
arr4=np.array(["Hello",1,3.14])
print(arr4,arr4.dtype)
arr5=np.array(["Hello",{1,2,3},3.14])
print(arr5,arr5.dtype)
arr6=np.array([5 + 6j])
print(arr6,arr6.dtype)
```
