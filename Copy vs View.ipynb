```python
import numpy as np

nums = np.array([1,2,3,4,5])

sub_nums = nums[0:3]

print(sub_nums)

sub_nums[1] = 200

print(sub_nums)

print(nums)
```

```python
import numpy as np

nums = np.array([1,2,3,4,5])

sub_nums = nums[0:3].copy()

print(sub_nums)

sub_nums[1] = 200

print(sub_nums)

print(nums)
```
