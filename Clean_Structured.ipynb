```python
import json
```

```python
#Loading the dataset
```

```python
def load_dataset(filename):
    with open(filename,"r") as f:
        data=json.load(f)
        return data 
```

```python
dataset=load_dataset("store_data.json")
print(dataset)
print(type(dataset))
```

```python
def clean_data(dataset):
    text_to_num={"one":1,"two":2,"three":3,"four":4,"five":5}
    duplicate_values=set()
    cleaned_data=[]
    for user in dataset:

        raw_ratings=str(user["rating"]).strip().lower()
        if(raw_ratings in text_to_num):
            raw_ratings=text_to_num[raw_ratings]
        user["rating"]=raw_ratings

        raw_age=user.get("age")
        if(raw_age==None):
            user["age"]=None

        if(user["name"].strip() in duplicate_values):
            continue
        duplicate_values.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data
            
        

        
        
```

```python
data=clean_data(dataset)
```

```python
print(data)
```

```python
def get_insights(data):
    total_avg_ratings=0
    poor_ratings=0

    for user in data:
        total_avg_ratings += float(user["rating"])
        avg=total_avg_ratings/len(data)

    for user in data:
        if(float(user["rating"]) < 3):
            poor_ratings += 1
            
        

    print(f"Average Ratings={avg}")
    print(f"Poor Ratings % is = {poor_ratings / len(data) * 100} %")
```

```python
get_insights(data)
```

```python
def get_recommendations(data):
    recommendations=[]
    current_recom={}

    for user in data:
        current_recom={}
        current_recom["user"]=user["name"]
        if(float(user["rating"]) >= 4):
            current_recom["Brand"]="Apple"
        else:
            current_recom["Brand"]="Samsung"

        recommendations.append(current_recom)
    return recommendations
            
```

```python
get_recommendations(data)
```
